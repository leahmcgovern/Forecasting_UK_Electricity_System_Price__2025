```python
def parse_date_sp_to_datetime(date_sp_str):
    date_part = date_sp_str[:-3]  # 'YYYY-MM-DD'
    period_part = int(date_sp_str[-2:])  # '01' to '48'

    # Base datetime object from YYYY-MM-DD
    dt_obj = pd.to_datetime(date_part)

    # Convert period number to actual hours and minutes
    # Period 1 = 00:00, Period 2 = 00:30, ..., Period 48 = 23:30
    hours = (period_part - 1) // 2
    minutes = ((period_part - 1) % 2) * 30

    return dt_obj + pd.Timedelta(hours=hours, minutes=minutes)

# Apply this conversion to the 'date_sp' column in the entire DataFrame
df['date_sp'] = df['date_sp'].apply(parse_date_sp_to_datetime)

print("DataFrame with 'date_sp' converted to datetime:")
display(df.head())
df.info()
```

```python
# Set date_sp as the index
df = df.set_index('date_sp').sort_index()

#Split into train and test using the index labels
train_V1_df = df.loc['2021-04-01':'2024-03-31'].copy()
test_V1_df = df.loc['2024-04-01':'2025-03-31'].copy()

# Save the split datasets to CSV files
train_V1_df.to_csv('train_transformed_V1_data.csv')
test_V1_df.to_csv('test_transformed_V1_data.csv')

# Verify the results
print(f"Training data saved: {train_V1_df.shape[0]} rows (from {train_V1_df.index.min()} to {train_V1_df.index.max()})")
print(f"Testing data saved: {test_V1_df.shape[0]} rows (from {test_V1_df.index.min()} to {test_V1_df.index.max()})")
```

```python
#plot transformed biomass

import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
sns.histplot(biomass, kde=True)
plt.title('Distribution of Log1p-transformed Biomass Generation')

plt.subplot(1, 3, 2)
sns.boxplot(x=biomass)
plt.title('Boxplot of Log1p-transformed Biomass Generation')

plt.subplot(1, 3, 3)
sm.qqplot(biomass, line='s', ax=plt.gca())
plt.title('QQ-plot of Log1p-transformed Biomass Generation')

plt.tight_layout()
plt.show()
```

```python
#pre-transformation plots

df_base.columns = df_base.columns.str.strip()
cols_base = [
      'temp_degcels', 'gasprice_ppmwh', 'marketindexprice_ppmwh',
       'sysprice_ppmwh', 'deratedmargin_mw', 'net_physnotif_mw',
       'dem_forecast_mw', 'gen_interconnector_mw', 'gen_biomass_mw',
       'gen_gas_mw', 'gen_pumped_mw', 'gen_riverhydro_mw', 'gen_nuclear_mw',
       'gen_solar_mw', 'gen_other_mw', 'gen_forecastwind_mw'
       ]

# Only loop through columns that actually exist in df_base to avoid KeyErrors
existing_cols = [c for c in cols_base if c in df_base.columns]

for col in existing_cols:
    plt.figure(figsize=(18, 5))

    # Histogram
    plt.subplot(1, 3, 1)
    sns.histplot(df_base[col], kde=True, color='skyblue')
    plt.title(f'Distribution of {col}')

    # Boxplot
    plt.subplot(1, 3, 2)
    sns.boxplot(x=df_base[col], color='lightgreen')
    plt.title(f'Boxplot of {col}')

    # QQ-plot
    plt.subplot(1, 3, 3)
    sm.qqplot(df_base[col].dropna(), line='s', ax=plt.gca())
    plt.title(f'QQ-plot of {col}')

    plt.tight_layout()
    plt.show()

# Print a warning for anything that was missed
missing = set(cols_base) - set(existing_cols)
if missing:
    print(f"Note: The following columns were not found: {missing}")
```
