```python
date_df = pd.read_csv("/content/drive/MyDrive/UCL/Energy Data Analysis Group 3 Coursework/data/Merged/base_merge_file.csv")
```

```python
base_df = date_df
for csv_file in csv_files:
      file_path = os.path.join(SHARED_DRIVE_PATH, csv_file)

      try:
          df = pd.read_csv(file_path)
          df.columns = df.columns.str.lower()

          for x in naming_rules:
            if x == csv_file:
              print(f"Processing {csv_file}\nthe names of the columns for the file being merged are:\n{df.columns}")
              base_df = pd.merge(base_df,df,"left",on="date_sp")
              print(f"Success!!! \nThe merged column names are:\n{base_df.columns}\n")

      except Exception as e:
          print(f"\nError reading {csv_file}: \n {e}")
```

```python
merged_file_save_location = "/content/drive/MyDrive/UCL/Energy Data Analysis Group 3 Coursework/data/Merged/merged_data.csv"

base_df.to_csv(merged_file_save_location, index=False)
```
