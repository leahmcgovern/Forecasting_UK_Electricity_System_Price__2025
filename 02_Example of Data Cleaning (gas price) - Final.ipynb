```python
NG_SAP_Hourly_Clean = NG_SAP_Hourly_Raw.rename(columns={'Applicable At': 'Data Created',
                  'Applicable For': 'Gas Day', 'Data Item': 'Label','Value':'Value (£/MWh)'},inplace=False)
NG_SAP_Hourly_Clean
```

```python
NG_SAP_Hourly_Clean['Value (£/MWh)'] = NG_SAP_Hourly_Clean['Value (£/MWh)']*10
NG_SAP_Hourly_Clean
```

```python
#spltting up date and time column
def split_datetime_column(df, column_name):
    df = df.copy()

    df['Date'] = df[column_name].str.split(' ').str[0]
    df['Time'] = df[column_name].str.split(' ').str[1]

    return df

NG_SAP_Hourly_processed = split_datetime_column(
    NG_SAP_Hourly_Clean,
    'Data Created')

NG_SAP_Hourly_Clean_processed
```

```python
adj_columns = ['Date', 'Time', 'Label', 'Value (£/MWh)']
NG_SAP_Hourly_adj = NG_SAP_Hourly_Clean_processed[adj_columns].copy()
NG_SAP_Hourly_adj
```

```python
#convert to datetime object
NG_SAP_Hourly_adj['Date'] = pd.to_datetime(NG_SAP_Hourly_adj['Date'], format='%d/%m/%Y')

#adjustformat to be by YYYY_MM_DD
NG_SAP_Hourly_adj['Date'] = NG_SAP_Hourly_adj['Date'].dt.strftime('%Y-%m-%d')

NG_SAP_Hourly_adj
```

```python
# need to recombine date and time to round the time to nearest hour
NG_SAP_Hourly_adj['Datetime_Full'] = pd.to_datetime(
    NG_SAP_Hourly_adj['Date'] + ' ' + NG_SAP_Hourly_adj['Time'])

NG_SAP_Hourly_adj['Time_Rounded'] = NG_SAP_Hourly_adj['Datetime_Full'].dt.round('h')

# 3. Display the first few rows to confirm the result
NG_SAP_Hourly_adj
```

```python
new_columns = ['Time_Rounded', 'Label', 'Value (£/MWh)']
NG_SAP_Hourly_adj2 = NG_SAP_Hourly_adj[new_columns].copy()
NG_SAP_Hourly_adj2
```

```python
#if there were duplicates because of when hourly readings were taken, cleans them and takes mean based on Value column
NG_SAP_Hourly_adj2 = NG_SAP_Hourly_adj2.groupby('Time_Rounded')['Value (£/MWh)'].mean().reset_index()

# set index
NG_SAP_Hourly_adj2 = NG_SAP_Hourly_adj2.set_index('Time_Rounded')

#set data to 30-minute intervals
NG_SAP_Hourly_resampled = NG_SAP_Hourly_adj2.reindex(
    pd.date_range(
        start=NG_SAP_Hourly_adj2.index.min(),
        end=NG_SAP_Hourly_adj2.index.max(),
        freq='30min'))

# 4. Fill the new NaN rows (the half-hour points) with the last known value
NG_SAP_Hourly_resampled['Value (£/MWh)'] = NG_SAP_Hourly_resampled['Value (£/MWh)'].ffill()

NG_SAP_Hourly_resampled = NG_SAP_Hourly_resampled.rename_axis('Date_Time')
# Display the result
NG_SAP_Hourly_resampled
```

```python
NG_SAP_Hourly_resampled = NG_SAP_Hourly_resampled.rename(columns={'Value (£/MWh)':'Gas Price (£/MWh)'})
NG_SAP_Hourly_resampled
```

```python
#just double checkign for duplicates
#with leapyear I think number of inputs should be 70,128
#mine includes most of 4/1 so I think that's why I"m over

is_duplicate = NG_SAP_Hourly_resampled.index.duplicated().any()
print(f"Are there any duplicate Date_Time entries? {is_duplicate}")
```

```python
#now I want to change time to settlement period

index_dt = NG_SAP_Hourly_resampled.index

#set up calc for SP:
#Calculate the total minutes past midnight (e.g., 01:30 is 90 minutes)
# Then divide by 30 minutes and add 1 to get the SP (1 to 48)
# Example: 00:00 is (0/30) + 1 = SP 1
# Example: 00:30 is (30/30) + 1 = SP 2
# Example: 23:30 is (1410/30) + 1 = SP 48
settlement_period = ((index_dt.hour * 60 + index_dt.minute) / 30).astype(int) + 1

#leading zero for SP
sp_formatted = pd.Series(settlement_period).astype(str).str.zfill(2).values
#format to add SP to date
date_part = index_dt.strftime('%Y-%m-%d')
new_index_values = date_part + '-' + sp_formatted

NG_SAP_Hourly_resampled.index = new_index_values

NG_SAP_Hourly_resampled
```

```python
Gas_Price_Final = NG_SAP_Hourly_resampled.rename_axis('Date_sp')
```
